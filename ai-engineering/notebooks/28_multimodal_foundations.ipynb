## Step 1 — Load bundled images

The digits dataset ships with sklearn, so no download is needed. Images are 8×8 grayscale arrays, flattened for a simple classifier.


```python
import numpy as np
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
digits=load_digits()
Xtr,Xte,ytr,yte=train_test_split(digits.data,digits.target,stratify=digits.target,random_state=42)
assert digits.images.shape[1:]==(8,8)
print('Images:',digits.images.shape,'pixel range:',digits.data.min(),digits.data.max())
```

**Check your result:** There are 8×8 grayscale images with pixel values from 0 to 16.


## Step 2 — Train and evaluate vision features

This linear classifier is a baseline. CNNs and vision transformers learn representations that account for richer spatial structure.


```python
model=make_pipeline(StandardScaler(),LogisticRegression(max_iter=1000)).fit(Xtr,ytr)
accuracy=accuracy_score(yte,model.predict(Xte))
assert accuracy>0.9
print('Digit test accuracy:',accuracy)
```

**Check your result:** Accuracy exceeds 90% on this small bundled dataset.


## Step 3 — Extract audio features

Generate a one-second 440 Hz sine wave. An FFT finds frequency energy; this is signal processing, not speech recognition.


```python
sample_rate=8000
t=np.arange(sample_rate)/sample_rate
audio=np.sin(2*np.pi*440*t)
spectrum=np.abs(np.fft.rfft(audio))
frequencies=np.fft.rfftfreq(len(audio),1/sample_rate)
peak=float(frequencies[np.argmax(spectrum)])
assert abs(peak-440)<1
print('Peak frequency Hz:',peak,'RMS:',np.sqrt(np.mean(audio**2)))
```

**Check your result:** The frequency peak is 440 Hz.
