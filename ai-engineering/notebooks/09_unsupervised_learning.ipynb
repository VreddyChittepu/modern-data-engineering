## Step 1 — Create unlabeled examples

The generated group labels are used only for understanding, not for fitting.


```python
import numpy as np
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
from sklearn.ensemble import IsolationForest
X,_ = make_blobs(n_samples=400,centers=3,n_features=5,cluster_std=1,random_state=42)
Z = StandardScaler().fit_transform(X)
print(Z.shape)
```

**Check your result:** Four hundred rows with five standardized features.


## Step 2 — Cluster and compress

Silhouette compares within-cluster tightness to separation. PCA finds high-variance linear directions.


```python
cluster = KMeans(n_clusters=3,n_init=10,random_state=42).fit(Z)
pca = PCA(n_components=2).fit(Z)
embedding = pca.transform(Z)
print('Silhouette:',silhouette_score(Z,cluster.labels_))
print('PCA variance retained:',pca.explained_variance_ratio_.sum())
assert embedding.shape == (400,2)
```

**Check your result:** The two-dimensional representation has one row per example.


## Step 3 — Detect unusual points

IsolationForest returns −1 for anomalies. An outlier flag is a review signal, not proof of an error.


```python
anomaly_model = IsolationForest(contamination=0.03,random_state=42).fit(Z)
flags = anomaly_model.predict(Z)
print('Flagged rows:',int((flags == -1).sum()))
assert set(flags).issubset({-1,1})
```

**Check your result:** Approximately 3% of rows are flagged.
