## Step 3 — Run a smoke check

This checks CPU numerical computation. The course uses NumPy arrays and Spark DataFrames directly; there is no pandas conversion.


```python
import numpy as np
from sklearn.linear_model import LinearRegression
X = np.arange(10, dtype=float).reshape(-1, 1)
y = 3 * X[:, 0] + 2
model = LinearRegression().fit(X, y)
assert np.allclose(model.predict([[10.0]]), [32.0])
print('Core setup passed')
```

**Check your result:** Core setup passed.
