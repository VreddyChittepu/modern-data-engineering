## Step 1 — Create a feature matrix

Rows represent examples; columns represent features. A model expects a consistent schema.


```python
import numpy as np
rng = np.random.default_rng(42)
X = rng.normal(size=(8, 3))
assert X.shape == (8, 3)
print('shape:', X.shape, 'column means:', X.mean(axis=0))
```

**Check your result:** Eight examples, three features.


## Step 2 — Handle missing values

Fit imputation statistics on training data only. This tiny example explicitly separates training and scoring rows.


```python
train = np.array([[1., 10.], [3., np.nan], [5., 30.]])
score = np.array([[np.nan, 50.]])
means = np.nanmean(train, axis=0)
def impute(values, fitted_means):
    return np.where(np.isnan(values), fitted_means, values)
clean = impute(train, means)
assert not np.isnan(clean).any()
assert impute(score, means)[0, 0] == 3.0
print(clean)
```

**Check your result:** The missing training value becomes 20; the scoring row uses the training mean 3.


## Step 3 — Vectorize a prediction

Broadcasting avoids a Python loop over rows. Always check dimensions before multiplying.


```python
weights = np.array([0.2, -0.1, 0.5])
scores = X @ weights + 0.3
assert scores.shape == (8,)
loop_scores = np.array([sum(row * weights) + 0.3 for row in X])
assert np.allclose(scores, loop_scores)
print(scores[:3])
```

**Check your result:** The vectorized and loop results match.
