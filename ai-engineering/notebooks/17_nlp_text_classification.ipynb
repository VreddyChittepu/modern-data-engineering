## What you will learn

Create a sparse TF-IDF representation and train a simple intent classifier.


## Step 1 — Create labeled text

Tiny hand-authored examples make the mechanism visible. Real evaluation needs a much larger representative set.


```python
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
texts=['refund my order','return the item','please refund this purchase','cancel and refund',
       'where is my shipment','track my package','delivery status please','when will it arrive',
       'cannot sign in','reset my password','account login failed','forgot my password']
labels=['refund']*4+['delivery']*4+['login']*4
model=make_pipeline(TfidfVectorizer(ngram_range=(1,2)),LogisticRegression(C=5,max_iter=500))
model.fit(texts,labels)
print('Vocabulary size:',len(model[0].vocabulary_))
```

**Check your result:** The vectorizer learns words and bigrams only from training texts.


## Step 2 — Score held-out phrases

These phrases are excluded from fitting. A bag-of-words model has limited understanding of paraphrases and negation.


```python
test=['please reset password','track delivery','refund the item']
truth=['login','delivery','refund']
pred=model.predict(test)
print(classification_report(truth,pred,zero_division=0))
assert len(pred)==3
```

**Check your result:** One intent is returned per phrase.


## Step 3 — Inspect sparse features

Sparse matrices save memory because most documents contain only a small fraction of the vocabulary.


```python
matrix=model[0].transform(test)
print('Shape:',matrix.shape,'nonzero entries:',matrix.nnz)
print('Unknown vocabulary:',model[0].transform(['zqxw']).nnz)
assert model[0].transform(['zqxw']).nnz==0
```

**Check your result:** A completely unknown token produces an all-zero vector.
