## What you will learn

Understand dot products, MSE, derivatives, learning rate, and a numerical gradient check.


## Step 1 — Generate a linear relationship

We know the true slope, so we can check whether optimization recovers it despite noise.


```python
import numpy as np
rng = np.random.default_rng(42)
x = rng.uniform(-1, 1, 300)
y = 2.5 * x - 0.7 + rng.normal(0, 0.05, len(x))
X = np.column_stack([np.ones(len(x)), x])
w = np.zeros(2)
def loss(w):
    return np.mean((X @ w - y) ** 2)
print('Initial MSE:', loss(w))
```

**Check your result:** MSE is positive because zero weights cannot explain the target.


## Step 2 — Verify the derivative

A finite difference check catches sign and scaling mistakes before training.


```python
gradient = 2 * X.T @ (X @ w - y) / len(y)
eps = 1e-6
numeric = np.array([(loss(w + eps * np.eye(2)[i]) - loss(w - eps * np.eye(2)[i])) / (2 * eps) for i in range(2)])
assert np.allclose(gradient, numeric, atol=1e-6)
print('Gradient check passed')
```

**Check your result:** Analytical and numerical gradients agree.


## Step 3 — Optimize

Each update moves against the gradient. A very large learning rate can make the loss diverge.


```python
history = []
for epoch in range(400):
    gradient = 2 * X.T @ (X @ w - y) / len(y)
    w -= 0.1 * gradient
    history.append(loss(w))
assert history[-1] < history[0]
assert np.allclose(w, [-0.7, 2.5], atol=0.05)
print('intercept, slope:', w, 'final MSE:', history[-1])
```

**Check your result:** Weights are close to intercept −0.7 and slope 2.5.


## Step 4 — Connect logits to probability

The sigmoid maps any real score into (0,1); it does not guarantee probability calibration.


```python
logits = np.array([-2., 0., 2.])
probabilities = 1 / (1 + np.exp(-logits))
assert probabilities[1] == 0.5
print(probabilities)
```

**Check your result:** The probabilities increase as the logits increase.
