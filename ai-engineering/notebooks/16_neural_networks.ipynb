## What you will learn

Train a two-layer network on XOR; learn activations, BCE, gradients, and weight updates.


## Step 1 — Initialize the network

XOR cannot be separated by one straight line. A hidden nonlinear layer can represent it. Biases and weights are both trainable.


```python
import numpy as np
rng=np.random.default_rng(42)
X=np.array([[0.,0.],[0.,1.],[1.,0.],[1.,1.]])
y=np.array([[0.],[1.],[1.],[0.]])
W1=rng.normal(0,1,(2,8)); b1=np.zeros((1,8))
W2=rng.normal(0,1,(8,1)); b2=np.zeros((1,1))
def sigmoid(z): return 1/(1+np.exp(-np.clip(z,-40,40)))
def forward(X):
    hidden=np.tanh(X@W1+b1)
    return hidden,sigmoid(hidden@W2+b2)
print('Initial probabilities:',forward(X)[1].ravel())
```

**Check your result:** Four probabilities are produced, one per XOR input.


## Step 2 — Train using backpropagation

For sigmoid plus binary cross entropy, the output gradient is (p-y)/n. Update all parameters after calculating every gradient.


```python
losses=[]
for epoch in range(4000):
    h,p=forward(X)
    losses.append(float(-np.mean(y*np.log(p+1e-9)+(1-y)*np.log(1-p+1e-9))))
    dz=(p-y)/len(X)
    dW2=h.T@dz; db2=dz.sum(axis=0,keepdims=True)
    dh=(dz@W2.T)*(1-h*h)
    dW1=X.T@dh; db1=dh.sum(axis=0,keepdims=True)
    W1-=0.3*dW1; b1-=0.3*db1; W2-=0.3*dW2; b2-=0.3*db2
print('Initial/final BCE:',losses[0],losses[-1])
assert losses[-1]<0.05
```

**Check your result:** Loss decreases substantially.


## Step 3 — Verify predictions

This checks learning the truth table, not generalization to a new distribution.


```python
pred=(forward(X)[1]>=0.5).astype(int)
assert np.array_equal(pred,y.astype(int))
print(pred.ravel())
```

**Check your result:** Predictions are [0, 1, 1, 0].
